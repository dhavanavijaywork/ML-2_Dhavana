```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("enjoysport.csv")
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
h = np.array(["Ø"] * X.shape[1])
```

```python
for x, target in zip(X, y):
    if target == "Yes":
        h = np.where(h == "Ø", x,
             np.where(h == x, h, "?"))
```

```python
print("Final Hypothesis:", h)
```
